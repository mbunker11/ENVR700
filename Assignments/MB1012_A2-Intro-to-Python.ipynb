````python
# Write your solution here

print("Function with One Input")
import math

def circle_area(radius):
    circle_area = radius*radius*math.pi
    return(circle_area)

circle_area(7)


````

````python
print("Fucntion with Two Inputs")

def rectangle_area(length, width):
    area = length*width
    return(area)

rectangle_area(7,8)
````
